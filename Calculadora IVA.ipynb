#Calculadora de IVA


```python
#--------------------------------------------#
# esta aplicacion calcula el IVA a el precio #
# de un producto a la venta                  #
# la variable de entrada es:                 #
# int_valor                                  #
# variable interna:                          #
# f_impuesto                                 #
# variable de salida:                        #
# f_iva                                      #
#--------------------------------------------#
int_valor = 0
f_impuesto = 0.16
f_iva = 0.0
#---------- Entrada de datos ----------------#
int_valor = int (input("Ingrese el valor del producto: "))
#---------- Calculos ------------------------#
f_iva = int_valor * f_impuesto
print("El I.V.A. del articulo es:"+str(f_iva))
print("El valor total es:"+str(int_valor+f_iva))
```
